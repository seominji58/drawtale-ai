## 2. dev 코드에 우리 모델 연결 (포즈 모델 선택 추가)

````python
%%writefile /content/drawtale-dev/runtime/ai_server/a2_pose.py
"""A2 포즈 모델 (RTMPose-m drawing15, ONNX) — analyzer.estimate_pose 와 같은 입출력.

MMPose 1.x 추론(inference_topdown, 잘라 낸 그림 전체를 사람 영역으로)을 numpy로 그대로 재현한다.
  사람 영역 = 그림 전체 × 1.25 → 가로세로 비율 192:256 맞춤 → 아핀 변환 → RGB 정규화
  → 원본 + 좌우 뒤집은 입력을 함께 추론, 뒤집은 쪽은 좌우 관절을 바꿔 평균 (flip test)
  → SimCC 최댓값 위치 / 2 → 원래 좌표로 되돌림
반환 관절 이름·규칙은 Meta 경로(estimate_pose)와 같다: neck = 코, torso = 양 어깨 가운데, hip = 양 엉덩이 가운데.
"""
import os

import cv2
import numpy as np

ONNX_PATH = os.getenv("A2_ONNX", "/content/models/rtmpose-m_drawing15_pre.onnx")
IN_W, IN_H, SPLIT, PADDING = 192, 256, 2.0, 1.25
MEAN = np.array([123.675, 116.28, 103.53], np.float32)
STD = np.array([58.395, 57.12, 57.375], np.float32)
# 우리 15관절 순서: 0 nose 1 neck 2 r_sho 3 r_elb 4 r_wri 5 l_sho 6 l_elb 7 l_wri 8 pelvis 9 r_hip 10 r_knee 11 r_ank 12 l_hip 13 l_knee 14 l_ank
FLIP = [0, 1, 5, 6, 7, 2, 3, 4, 8, 12, 13, 14, 9, 10, 11]
# 런타임 관절 이름 ← 우리 관절 번호 (estimate_pose 의 순서 그대로)
RUNTIME = [("hip", 8), ("torso", 1), ("neck", 0),
           ("right_shoulder", 2), ("right_elbow", 3), ("right_hand", 4),
           ("left_shoulder", 5), ("left_elbow", 6), ("left_hand", 7),
           ("right_hip", 9), ("right_knee", 10), ("right_foot", 11),
           ("left_hip", 12), ("left_knee", 13), ("left_foot", 14)]
_SESSION = None


def _session():
    global _SESSION
    if _SESSION is None:
        import onnxruntime as ort
        _SESSION = ort.InferenceSession(ONNX_PATH, providers=["CPUExecutionProvider"])
    return _SESSION


def _third(a, b):
    d = a - b
    return b + np.r_[-d[1], d[0]]


def _warp_matrix(center, scale):
    """mmpose get_warp_matrix (회전 0, shift 0, 비율 고정)"""
    src = np.zeros((3, 2), np.float32); dst = np.zeros((3, 2), np.float32)
    src[0] = center; src[1] = center + np.array([scale[0] * -0.5, 0.])
    dst[0] = [IN_W * 0.5, IN_H * 0.5]; dst[1] = dst[0] + np.array([IN_W * -0.5, 0.])
    src[2] = _third(src[0], src[1]); dst[2] = _third(dst[0], dst[1])
    return cv2.getAffineTransform(np.float32(src), np.float32(dst))


def _simcc_max(sx, sy):
    """mmpose get_simcc_maximum: 위치 = 최댓값 칸, 점수 = x·y 최댓값 중 작은 쪽"""
    locs = np.stack([sx.argmax(1), sy.argmax(1)], -1).astype(np.float32)
    vals = np.minimum(sx.max(1), sy.max(1))
    locs[vals <= 0.] = -1
    return locs, vals


def infer(img_bgr):
    """잘라 낸 그림(BGR) 전체를 사람 영역으로 → (15, 2) 좌표(그림 기준), (15,) 점수"""
    h, w = img_bgr.shape[:2]
    center = np.array([w * 0.5, h * 0.5], np.float32)
    sw, sh = w * PADDING, h * PADDING
    aspect = IN_W / IN_H
    scale = np.array([sw, sw / aspect] if sw > sh * aspect else [sh * aspect, sh], np.float32)
    x = cv2.warpAffine(img_bgr, _warp_matrix(center, scale), (IN_W, IN_H), flags=cv2.INTER_LINEAR)
    x = (cv2.cvtColor(x, cv2.COLOR_BGR2RGB).astype(np.float32) - MEAN) / STD
    x = x.transpose(2, 0, 1)
    batch = np.stack([x, x[:, :, ::-1]]).astype(np.float32)            # 원본 + 좌우 뒤집기 (inputs.flip(-1)과 같음)
    sx, sy = _session().run(None, {"input": batch})
    px = (sx[0] + sx[1][FLIP][:, ::-1]) * 0.5                          # flip_vectors: 좌우 관절 바꾸고 x축 뒤집기
    py = (sy[0] + sy[1][FLIP]) * 0.5
    locs, vals = _simcc_max(px, py)
    kp = locs / SPLIT / np.array([IN_W, IN_H], np.float32) * scale + center - 0.5 * scale
    return kp, vals


def estimate_pose_a2(cropped, bbox):
    """analyzer.estimate_pose 와 같은 형식: [{name, x, y, score}] × 15 (좌표는 분석 이미지 기준)"""
    kp, vals = infer(cropped)
    return [{"name": name,
             "x": round(float(kp[i, 0]) + bbox["left"]),
             "y": round(float(kp[i, 1]) + bbox["top"]),
             "score": round(min(1.0, max(0.0, float(vals[i]))), 3)}
            for name, i in RUNTIME]
````

````python
%%writefile /content/patch_analyzer.py
"""dev analyzer.py 에 포즈 모델 선택 추가: POSE_PROVIDER=a2 이면 우리 모델(ONNX), 아니면 Meta (기존 그대로)"""
import sys
from pathlib import Path
p = Path(sys.argv[1]) / "runtime/ai_server/analyzer.py"
s = p.read_text(encoding="utf-8")
old = """    # 3. Pose Estimator
    joints = estimate_pose(
        cropped,
        bbox
    )
"""
new = """    # 3. Pose Estimator (POSE_PROVIDER=a2 이면 A2 손그림 모델, 기본은 Meta)
    if os.getenv("POSE_PROVIDER", "meta").strip().lower() == "a2":
        from ai_server.a2_pose import estimate_pose_a2
        joints = estimate_pose_a2(cropped, bbox)
    else:
        joints = estimate_pose(
            cropped,
            bbox
        )
"""
if "POSE_PROVIDER" in s:
    print("이미 패치됨")
else:
    assert old in s, "analyzer.py가 예상과 다름 (dev 브랜치가 바뀌었는지 확인)"
    s = s.replace(old, new, 1)
    if "\nimport os\n" not in s:
        s = "import os\n" + s
    p.write_text(s, encoding="utf-8")
    print("패치 완료")
````

### 6-2. 변환 + 확인

````python
%%writefile /content/export_onnx.py
"""우리 모델 → ONNX (입력 1장 또는 원본+뒤집기 2장, 출력 SimCC x·y). 런타임은 onnxruntime만으로 추론"""
import os, sys, torch
os.environ['MPLBACKEND'] = 'Agg'
from mmpose.apis import init_model
CFG, CK, OUT = sys.argv[1], sys.argv[2], sys.argv[3]
m = init_model(CFG, CK, device='cpu').eval()

class Net(torch.nn.Module):
    def __init__(self, m):
        super().__init__(); self.m = m
    def forward(self, x):
        return self.m.head(self.m.extract_feat(x))          # (simcc_x, simcc_y)

with torch.no_grad():
    sx, sy = Net(m)(torch.zeros(2, 3, 256, 192))
    print('출력 크기', tuple(sx.shape), tuple(sy.shape))
    torch.onnx.export(Net(m), torch.zeros(2, 3, 256, 192), OUT, input_names=['input'], output_names=['simcc_x', 'simcc_y'],
                      dynamic_axes={'input': {0: 'n'}, 'simcc_x': {0: 'n'}, 'simcc_y': {0: 'n'}}, opset_version=17)
print('저장:', OUT, f'{os.path.getsize(OUT) / 1e6:.1f}MB')
````

````python
%%writefile /content/validate_onnx.py
"""ONNX + numpy 추론(런타임 a2_pose) 이 MMPose 추론(inference_topdown) 과 같은지 확인 — 같은 그림 전체를 사람 영역으로"""
import os, sys, glob, numpy as np, cv2
os.environ['MPLBACKEND'] = 'Agg'
from mmpose.apis import init_model, inference_topdown
CFG, CK, ONNX, PHOTOS = sys.argv[1:5]
os.environ['A2_ONNX'] = ONNX
sys.path.insert(0, '/content/drawtale-dev/runtime')
from ai_server import a2_pose
m = init_model(CFG, CK, device='cpu')
worst = 0
for f in sorted(glob.glob(f'{PHOTOS}/*'))[:6]:
    img = cv2.imread(f)
    s = 1000 / max(img.shape[:2])
    if s < 1:
        img = cv2.resize(img, None, fx=s, fy=s)
    r = inference_topdown(m, img)[0]
    k_ref, s_ref = r.pred_instances.keypoints[0], r.pred_instances.keypoint_scores[0]
    k, v = a2_pose.infer(img)
    d = np.abs(k - k_ref).max(); worst = max(worst, d)
    print(f'{os.path.basename(f)[:30]:30s} 좌표 최대 차이 {d:.2f}px | 점수 최대 차이 {np.abs(v - s_ref).max():.4f}')
print('결과:', '일치 (런타임 연결 OK)' if worst < 2 else '차이가 큼 → 확인 필요')
````

````python
# 같은 사진을 두 서버에 보내 dev 분석 결과(사람 영역, 관절 15개, 점수, 처리 시간) 저장
import json, time, requests
RES = {'meta': {}, 'a2': {}}
for n, f in enumerate(PHOTO_FILES):
    name = os.path.basename(f)
    for model, (port, _) in servers.items():
        t0 = time.time()
        with open(f, 'rb') as fh:
            r = requests.post(f'http://127.0.0.1:{port}/internal/v1/analyze', files={'file': (name, fh)}, timeout=600).json()
        r['wall_ms'] = round((time.time() - t0) * 1000)
        RES[model][name] = r
        if r.get('request_id'):
            requests.delete(f'http://127.0.0.1:{port}/internal/v1/sessions/{r["request_id"]}', timeout=30)
    print(f'{n + 1}/{len(PHOTO_FILES)} {name[:40]:40s} meta {"성공" if RES["meta"][name].get("success") else RES["meta"][name].get("message", "")[:40]} | '
          f'우리 {"성공" if RES["a2"][name].get("success") else RES["a2"][name].get("message", "")[:40]}', flush=True)
json.dump(RES, open(f'{OUT}/dev_compare_results.json', 'w'), ensure_ascii=False, indent=1)
print('저장:', f'{OUT}/dev_compare_results.json')
````

## 9. 결과
- **관절 위치 차이**: 두 모델이 같은 관절을 얼마나 다른 곳에 찍었나 (사람 영역 긴 변 대비). 정답이 없어서 "누가 맞나"는 아니고, 차이가 큰 관절·사진을 찾는 용도 → 10번 확인 이미지로 판단
- **분석 성공·실패**: dev는 관절 점수 평균이 0.35 미만이면 그림을 거절(`LOW_CONFIDENCE`). 모델마다 점수 크기가 달라 거절 수가 달라질 수 있음
- **처리 시간**: dev 서버의 분석 전체 시간. Meta는 TorchServe(HTTP)로, 우리 모델은 서버 안에서 ONNX로 돌아 방식 차이가 섞임

````python
# 결과 정리: 성공·실패, 두 모델의 관절 위치 차이(관절별), 점수, 처리 시간
import numpy as np, pandas as pd, json
RES = json.load(open(f'{OUT}/dev_compare_results.json'))
NAMES = ['hip', 'torso', 'neck', 'right_shoulder', 'right_elbow', 'right_hand', 'left_shoulder', 'left_elbow', 'left_hand',
         'right_hip', 'right_knee', 'right_foot', 'left_hip', 'left_knee', 'left_foot']
KO = dict(hip='골반', torso='목(어깨 가운데)', neck='코', right_shoulder='오른어깨', right_elbow='오른팔꿈치', right_hand='오른손목',
          left_shoulder='왼어깨', left_elbow='왼팔꿈치', left_hand='왼손목', right_hip='오른엉덩이', right_knee='오른무릎',
          right_foot='오른발목', left_hip='왼엉덩이', left_knee='왼무릎', left_foot='왼발목')
photos = list(RES['meta'])
ok = {m: [p for p in photos if RES[m][p].get('success')] for m in RES}
print(f'사진 {len(photos)}장 | 분석 성공: Meta {len(ok["meta"])}장, 우리 {len(ok["a2"])}장')
for m in RES:
    fails = pd.Series([RES[m][p].get('message', '').split(':')[0] for p in photos if not RES[m][p].get('success')]).value_counts()
    if len(fails):
        print(f'  {m} 실패 이유:', dict(fails))
both = [p for p in photos if p in ok['meta'] and p in ok['a2']]
same_box = sum(RES['meta'][p]['bbox'] == RES['a2'][p]['bbox'] for p in both)
print(f'둘 다 성공 {len(both)}장 (사람 영역이 같은 사진 {same_box}장 — 같은 검출 모델이라 모두 같아야 정상)')

def J(r):
    return {j['name']: (j['x'], j['y'], j['score'] if j['score'] is not None else np.nan) for j in r['joints']}
rows, per = [], {n: [] for n in NAMES}
for p in both:
    a, b, box = J(RES['meta'][p]), J(RES['a2'][p]), RES['meta'][p]['bbox']
    size = max(box['right'] - box['left'], box['bottom'] - box['top'])
    d = {n: np.hypot(a[n][0] - b[n][0], a[n][1] - b[n][1]) / size for n in NAMES}
    for n in NAMES:
        per[n].append(d[n])
    rows.append(dict(photo=p, mean_diff=np.mean(list(d.values())), max_joint=max(d, key=d.get), max_diff=max(d.values()),
                     meta_score=np.nanmean([a[n][2] for n in NAMES]), a2_score=np.nanmean([b[n][2] for n in NAMES]),
                     meta_ms=RES['meta'][p]['processing_time_ms'], a2_ms=RES['a2'][p]['processing_time_ms']))
df = pd.DataFrame(rows).sort_values('mean_diff', ascending=False)
print('\n관절별: 두 모델 위치 차이 (사람 영역 긴 변 대비)')
jt = pd.DataFrame({KO[n]: dict(평균_차이=np.mean(v), 차이_5퍼센트_이내=np.mean(np.array(v) < 0.05), 차이_10퍼센트_이내=np.mean(np.array(v) < 0.1))
                   for n, v in per.items()}).T.round(3)
display(jt.sort_values('평균_차이', ascending=False))
print('\n처리 시간 (서버 분석 전체, ms) · 관절 점수 평균 (모델마다 점수 크기가 달라 직접 비교 불가)')
display(pd.DataFrame({'Meta': [df.meta_ms.median(), df.meta_ms.mean(), df.meta_score.mean()],
                      '우리': [df.a2_ms.median(), df.a2_ms.mean(), df.a2_score.mean()]},
                     index=['처리 시간 중앙값', '처리 시간 평균', '관절 점수 평균']).round(2))
print('\n사진별 (두 모델이 가장 다르게 찍은 순)'); display(df.head(15).round(3))
jt.to_csv(f'{OUT}/joint_diff.csv'); df.to_csv(f'{OUT}/photo_diff.csv', index=False)
````

## 10. 확인 이미지

````python
# 확인 이미지: 왼쪽 Meta(빨강) / 오른쪽 우리(파랑), 두 모델이 가장 다르게 찍은 사진부터
import cv2
from IPython.display import Image as IImage, display
BONES = [('hip', 'torso'), ('torso', 'neck'), ('torso', 'right_shoulder'), ('right_shoulder', 'right_elbow'), ('right_elbow', 'right_hand'),
         ('torso', 'left_shoulder'), ('left_shoulder', 'left_elbow'), ('left_elbow', 'left_hand'), ('hip', 'right_hip'),
         ('right_hip', 'right_knee'), ('right_knee', 'right_foot'), ('hip', 'left_hip'), ('left_hip', 'left_knee'), ('left_knee', 'left_foot')]
def draw(img, r, col):
    v = img.copy(); j = {x['name']: (int(x['x']), int(x['y'])) for x in r['joints']}; t = max(2, max(img.shape[:2]) // 250)
    for a, b in BONES:
        cv2.line(v, j[a], j[b], col, t)
    for p in j.values():
        cv2.circle(v, p, t + 2, col, -1)
    b = r['bbox']; m = int(0.08 * max(b['right'] - b['left'], b['bottom'] - b['top']))
    return v[max(0, b['top'] - m):b['bottom'] + m, max(0, b['left'] - m):b['right'] + m]
tiles = []
for p in df.photo:
    img = cv2.imread(os.path.join(PHOTOS, p)); h = 300
    a, b = draw(img, RES['meta'][p], (40, 40, 230)), draw(img, RES['a2'][p], (230, 120, 30))
    a = cv2.resize(a, (int(a.shape[1] * h / a.shape[0]), h)); b = cv2.resize(b, (int(b.shape[1] * h / b.shape[0]), h))
    t = np.concatenate([a, np.full((h, 6, 3), 255, np.uint8), b], 1)
    cv2.putText(t, p[:28], (4, 16), cv2.FONT_HERSHEY_SIMPLEX, 0.45, (0, 0, 0), 1); tiles.append(t)
W = max(t.shape[1] for t in tiles); pad = lambda t: np.pad(t, ((0, 8), (0, W - t.shape[1]), (0, 0)), constant_values=255)
cv2.imwrite(f'{OUT}/dev_compare_sheet.jpg', np.concatenate([pad(t) for t in tiles], 0))
display(IImage(f'{OUT}/dev_compare_sheet.jpg'))
````
